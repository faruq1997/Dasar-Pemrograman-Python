Buatlah sebuah program Python interaktif yang bertugas sebagai kalkulator biaya kuliah mahasiswa berdasarkan jumlah SKS (Satuan Kredit Semester) yang diambil, sekaligus memberikan peringatan status akademik berdasarkan jumlah SKS tersebut.

```python
sks = 100000
```

```python
mhs_1 = 20
```

```python
total_biaya = sks * mhs_1
```

```python
print("Total yang perlu dibayar untuk semester ini: ",total_biaya)
```

```python
# misal, semester 1 minimal harus ambil 20 sks
semester_1 = 20
```

```python
if(mhs_1 >= semester_1):
  print("Lanjut")
else:
  print("Minimal ambil ",semester_1," sks")
```
